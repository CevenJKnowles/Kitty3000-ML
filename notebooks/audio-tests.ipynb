```python
import soundfile as sf
from pathlib import Path
```

```python
DATA_DIR = Path("/home/cjk/Dev/04_Lewagon/cats/260908 CatSound_originals_only/CatSound_originals")

print("exists:", DATA_DIR.exists())
print("is_dir:", DATA_DIR.is_dir())
print("top-level entries:", list(DATA_DIR.iterdir())[:5] if DATA_DIR.exists() else "N/A")

files = list(DATA_DIR.rglob("*.mp3"))
print("mp3 count:", len(files))

results = []
for file in files:
    with sf.SoundFile(file) as f:
        duration = len(f) / f.samplerate
    results.append((file.name, duration))

results.sort(key=lambda x: x[1])

shortest = results[0]
longest = results[-1]

print(f"Shortest: {shortest[0]} — {shortest[1]:.2f}s")
print(f"Longest: {longest[0]} — {longest[1]:.2f}s")
```
